```python
#polynomial regression
import numpy as np
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

#data
df = pd.DataFrame({
    "Experience":[1,2,3,4,5]
})
x=df
y= [10,20,50,90,150]
#polyfeature
poly = PolynomialFeatures(degree=2)
x_poly = poly.fit_transform(x)
print(x_poly)

#model
model = LinearRegression()
model.fit(x_poly,y)

#predict
input_data= int(input('Enter the experience:'))
new_data = pd.DataFrame({
    "Experience":[input_data]
})


#convert new data
u_new_data= poly.fit_transform(new_data)
predicted_model=model.predict(u_new_data)
print(predicted_model)

plt.plot(x_poly,y)
plt.show()
```

```python
#poly regression -- overfitting

import numpy as np
import pandas as pd
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt

df=pd.DataFrame({
        "lable":[1,2,3,4,5]
})
x=df
y= [10,20,50,90,150]

poly = PolynomialFeatures(degree=3)
x_poly = poly.fit_transform(x)
print(x_poly)

#model
model = LinearRegression()
model.fit(x_poly,y)

#predict
input_data= float(input('Enter the LABEL:'))
new_data = pd.DataFrame({
    "lable":[input_data]
})

#inputdata to poly
u_new_data= poly.fit_transform(new_data)

#prediction
predicted_model=model.predict(u_new_data)
print(predicted_model[0])
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

# Dataset
df = pd.DataFrame({
    "Marks": [10, 20, 30, 40, 50, 60, 70, 80],
    "Result": ['fail', 'fail', 'fail', 'pass', 'pass', 'pass', 'pass', 'pass']
})

# Label Encoding
le = LabelEncoder()
df['Result'] = le.fit_transform(df['Result'])

print("Encoded Data:")
print(df)

# Features and Target
x = df[['Marks']]
y = df['Result']

# Model
model = DecisionTreeClassifier(random_state=42)
model.fit(x, y)

# User Input
input_data = int(input("Enter the marks: "))

new_data = pd.DataFrame({
    "Marks": [input_data]
})

# Prediction (Encoded)
prediction = model.predict(new_data)

# Label Decoding
result = le.inverse_transform(prediction)

print("Encoded Prediction:", prediction[0])
print("Final Prediction:", result[0])
```

```python
#linear regression

import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    "experience":[1,2,3,4,5],
    "salary":[3,5,7,9,11]
}

df=pd.DataFrame(data)
x=df[['experience']]
y=df['salary']

model=LinearRegression()
model.fit(x,y)

new_data = pd.DataFrame({
    "experience":[6]
})

predicted_data=model.predict(new_data)
print(predicted_data[0])
```

```python
#evaluation matrix - > to evaluate that prediction is how much correct
import numpy as np
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score

#actual value
y_true=np.array([50,60,75,90,100])

#predicted value
y_pred = np.array=([48,59,72,88,105])

#mean absoutle error
mae=mean_absolute_error(y_true,y_pred)
print(mae)

#mean squared error
mse=mean_squared_error(y_true,y_pred)
print(mse)
#r2 score
r2=r2_score(y_true,y_pred)
print(r2)

#rmse
rmse= np.sqrt(mse)
print(rmse)
```

```python
#lasso and ridge regression
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_absolute_error

data={
    "size":[1000,1200,1500,1800,2000,2200,2500,2700,3000,3200],
    "bedrooms":[2,2,3,3,3,4,4,4,5,5],
    "age":[20,18,15,12,10,8,6,5,4,2],
    "price":[30,35,45,50,55,60,68,72,80,85]
}

df=pd.DataFrame(data)
print(df)

#split data into x and y
x=df[['size','bedrooms','age']]
y=df['price']

#train test split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

ridge_p = Ridge(alpha=1.0)
ridge_p.fit(x_train,y_train)

#prediction
ridge_pred=ridge_p.predict(x_test)
print(ridge_pred)

#mae
ridge_mae=mean_absolute_error(y_test,ridge_pred)
print(ridge_mae)
```

```python
#lasso regression
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_absolute_error

data={
    "size":[1000,1200,1500,1800,2000,2200,2500,2700,3000,3200],
    "bedrooms":[2,2,3,3,3,4,4,4,5,5],
    "age":[20,18,15,12,10,8,6,5,4,2],
    "price":[30,35,45,50,55,60,68,72,80,85]
}

df=pd.DataFrame(data)
print(df)

#split data into x and y
x=df[['size','bedrooms','age']]
y=df['price']

#train test split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)

#model
lasso_p = Lasso(alpha=1.0)
lasso_p.fit(x_train,y_train)

#prediction
lasso_pred=lasso_p.predict(x_test)
print(lasso_pred)   #-> intrcept+[(coef1*fea1)+(coef2*fea2)+(coef3*fea3)]

#mae
lasso_mae=mean_absolute_error(y_test,lasso_pred)
print(lasso_mae)

#inercept
print(lasso_p.intercept_)

#coefficient
print(lasso_p.coef_)

```
